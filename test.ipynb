```python
# learning special methiods to enable more flexible classes and objects
class By_PI:
    def __init__(self, a_value = None):
        self.a_value = a_value
    def __call__(self, x):
        return x * 3
    def __repr__(self):
        return f"Value is: {self.a_value}"

    def __add__(self, y):
        return self.a_value + y
    def __radd__(self, y):
        return self.a_value + y
    
a_slice = By_PI(7)
a_slice(3)
b = a_slice + 5
c = 5 + a_slice
print(c)
a_slice

```
